```python
n = int(input())
s = input().split()
arr = []
for i in s:
    arr.append(int(i))
k = int(input())
if k > n:
    print("Invalid")
    exit()
windsum = 0
# first window
for i in range(k):
    windsum += arr[i]
maxsum = windsum
# sliding window
for i in range(k, n):
    windsum -= arr[i - k]
    windsum += arr[i]
    maxsum = max(maxsum, windsum)
print(maxsum)
```

```python
def isPalindrome(s, i, j):
    while i < j:
        if s[i] != s[j]:
            return False
        i += 1
        j -= 1
    return True
def minCuts(s, start):
    # if whole substring is palindrome or empty
    if start == len(s) or isPalindrome(s, start, len(s)-1):
        return 0

    ans = float('inf')

    for end in range(start, len(s)):
        if isPalindrome(s, start, end):
            # 1 cut + cuts for remaining string
            ans = min(ans, 1 + minCuts(s, end + 1))

    return ans
s = input().strip()
print(minCuts(s, 0))
```

```python
n,k=map(int, input().split())
arr=list(map(int, input().split()))
freq={}
result=[]
for i in range(k):
        if arr[i] in freq:
                freq[arr[i]]+=1
        else:
                freq[arr[i]]=1
result.append(len(freq))
for i in range(k,n):
        left=arr[i-k]
        freq[left]-=1
        if freq[left]==0:
                del freq[left]
        right=arr[i]
        if right in freq:
                freq[right]+=1
        else:
                freq[right]=1
        result.append(len(freq))
print(*result)
```
